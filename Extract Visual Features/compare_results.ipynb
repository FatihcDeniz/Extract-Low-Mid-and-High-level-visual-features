# Fourier Slope/Sigma

```python
import os
import time
import pandas as pd
from PIL import Image
import numpy as np

# Import your functions
from utils.AT.fourier_qips import fourier_redies as fourier_qips_func
from utils.AT.fourier_modified import fourier_redies as fourier_modified_func

# Folder with input images
image_folder = r"C:\Users\20225378\Desktop\IDEAS final\Images\bedroom"
n_images = 10
# Store results
results = []

# Loop through images
for filename in os.listdir(image_folder)[:n_images]:
    if filename.lower().endswith((".png", ".jpg", ".jpeg", ".bmp", ".tiff")):
        file_path = os.path.join(image_folder, filename)

        # Load image
        img = Image.open(file_path).convert("L")
        img_array = np.array(img)

        # ---- First function ----
        start_time = time.time()
        result_qips = fourier_qips_func(img_array)
        time_qips = time.time() - start_time

        # ---- Second function ----
        start_time = time.time()
        result_modified = fourier_modified_func(img_array)
        time_modified = time.time() - start_time

        # Store results
        results.append({
            "image": filename,
            "result_qips": result_qips,
            "time_qips_sec": time_qips,
            "result_modified": result_modified,
            "time_modified_sec": time_modified
        })

        print(f"Processed {filename}")

# Convert to DataFrame
df = pd.DataFrame(results)

print(f"Mean time original {df['time_qips_sec'].mean()} -- Mean time modified {df['time_modified_sec'].mean()}")
df
```

#### CNN symmetry comparison

```python
import os
import time
import pandas as pd
from PIL import Image
import numpy as np

# Import your functions
from utils.AT.CNN_qips import CNN_symmetry as symmetry_original
from utils.AT.CNN_modified import CNN_symmetry as symmetry_modified
n_images = 10
# Folder with input images
image_folder = r"C:\Users\20225378\Desktop\IDEAS final\Images\bedroom"
[kernel,bias] = np.load(open(r"utils\AT\bvlc_alexnet_conv1.npy", "rb"), encoding="latin1", allow_pickle=True)

# Store results
results = []


# Loop through images
for filename in os.listdir(image_folder)[:n_images]:
    if filename.lower().endswith((".png", ".jpg", ".jpeg", ".bmp", ".tiff")):
        file_path = os.path.join(image_folder, filename)

        # Load image
        img = Image.open(file_path).resize((224,224))
        img_array = np.array(img)

        # ---- First function ----
        start_time = time.time()
        result_qips = symmetry_original(img_array, kernel, bias)
        time_qips = time.time() - start_time

        # ---- Second function ----
        start_time = time.time()
        result_modified = symmetry_modified(img_array, kernel, bias)
        time_modified = time.time() - start_time

        # Store results
        results.append({
            "image": filename,
            "result_qips": result_qips,
            "time_qips_sec": time_qips,
            "result_modified": result_modified,
            "time_modified_sec": time_modified
        })

        print(f"Processed {filename}")

# Convert to DataFrame
df = pd.DataFrame(results)

print(f"Mean time original {df['time_qips_sec'].mean()} -- Mean time modified {df['time_modified_sec'].mean()}")
df
```
